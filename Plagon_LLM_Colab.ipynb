## 2. Mount Google Drive (Recommended)
Connect to Google Drive to save your trained model checkpoints automatically. This prevents data loss if the Colab session disconnects.

```python
from google.colab import drive
import os
import shutil

# Mount Google Drive
drive.mount('/content/drive')

# Define the path in Google Drive where checkpoints will be saved
drive_checkpoint_path = '/content/drive/MyDrive/Plagon_Checkpoints'

# Create the directory if it doesn't exist
os.makedirs(drive_checkpoint_path, exist_ok=True)

# Output current status
print(f"✅ Google Drive Mounted. Checkpoints will be saved to: {drive_checkpoint_path}")

# Link the project's 'checkpoints' folder to Google Drive
# This ensures that whenever the training script saves a model, it goes directly to Drive
if os.path.exists('checkpoints'):
    # Remove existing local folder if it's empty or you want to assist linking (be careful not to delete wanted data, but usually in fresh colab it's fine)
    # For safety in Colab (which is usually fresh), we remove the local dir to replace with symlink
    if not os.path.islink('checkpoints'):
        shutil.rmtree('checkpoints')

# Create symbolic link: ./checkpoints -> /content/drive/MyDrive/Plagon_Checkpoints
if not os.path.exists('checkpoints'):
    os.symlink(drive_checkpoint_path, 'checkpoints')
    print("✅ Symlink created. Local 'checkpoints' folder is now mapped to Google Drive.")
else:
    print("ℹ️ Checkpoints folder already exists or linked.")
```
